```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Dataset Load Karein
# (Make sure aapki file ka path yahi ho jo data/raw/ ke andar hai)
df = pd.read_csv('../data/raw/house_prices.csv')

# 2. Pehle 5 rows dekhein ki data kaisa dikhta hai
print("--- FIRST 5 ROWS ---")
display(df.head())

# 3. Data ki basic info check karein (Columns, Data Types, Missing Values)
print("\n--- DATA INFO ---")
df.info()

# 4. Statistical summary dekhein (Mean, Min, Max, etc.)
print("\n--- STATISTICAL SUMMARY ---")
display(df.describe())
```

```python
# 1. Price ka Distribution dekhein (Histogram)
plt.figure(figsize=(8, 4))
sns.histplot(df['Price'], kde=True, color='blue')
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()

# 2. Area aur Price ke beech ka relation (Scatter Plot)
plt.figure(figsize=(8, 4))
sns.scatterplot(x='Area', y='Price', hue='Location', data=df, palette='viridis')
plt.title('Area vs Price by Location')
plt.xlabel('Area (sqft)')
plt.ylabel('Price')
plt.show()

# 3. Correlation Matrix (Numeric features ke beech ka sambandh)
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
